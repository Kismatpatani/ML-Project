 What is the relationship between Credit Score and Loan Amount?

```python
# relationship between two numerical variables.
plt.figure(figsize=(12,6))

plt.scatter(
    df['CreditScore'],
    df['LoanAmount'],
    c=df['LoanAmount'],
    cmap='plasma'
)

plt.colorbar(label='Loan Amount')

plt.title("Credit Score vs Loan Amount")
plt.xlabel("Credit Score")
plt.ylabel("Loan Amount")

plt.show()
```

 What is the distribution of customers' Income?

```python
plt.figure(figsize=(12,5))

plt.hist(df['Income'], bins=10, color='purple', edgecolor='white')

plt.title("Income Distribution")
plt.xlabel("Income")
plt.ylabel("Frequency")

plt.show()
```
