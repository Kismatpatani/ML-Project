Step 7: Check Data Types

```python
df.dtypes
df["LoanAmount"] = df["LoanAmount"].astype(float)
print(df.dtypes)
```

Step 8: Detect Outliers

```python
plt.figure(figsize=(6,4))
sns.boxplot(x=df["LoanAmount"])
plt.title("Loan Amount Box Plot")
plt.show()
```

Step 9: Encode Categorical Variables


```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df["MaritalStatus"] = le.fit_transform(df["MaritalStatus"])

df["Education"] = le.fit_transform(df["Education"])
//convert catogarical values to numerical value
```

Step 10: Feature Scaling

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df[["Income","LoanAmount"]] = scaler.fit_transform(
    df[["Income","LoanAmount"]]
)
//scale (normalize) numeric data so that all values lie between 0 and 1.
```

Step 11: Clean Column Names

```python
df.columns = df.columns.str.strip().str.lower().str.replace(" ","_")
df.columns
```

Step 12: Data Visualization
